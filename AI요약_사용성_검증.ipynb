# AI 요약 화면의 버튼 활성화 조건 실습

Streamlit 없이 버튼 활성화 판단 함수만 검증합니다. 각 실습 아래 별도의 `assert` 검증 셀이 있습니다.


```python
from ui_helpers import ai_summary_blockers

# 1) PDF와 논문이 준비되어도 OpenAI API 키가 없으면 요약할 수 없습니다.
paper = {"자료명/저자사항": "데이터베이스 포렌식 논문"}
evidence = "이 연구는 데이터베이스 포렌식 기법을 분석합니다. " * 4
blockers = ai_summary_blockers(paper, True, evidence, "")
print("활성화 차단 원인:", blockers)
```

```python
# 검증 셀 1: 개인 API 키가 없어 버튼 비활성화
assert blockers == ["API_KEY"]
print("검증 1 통과: 개인 API 키를 입력해야 합니다.")
```

```python
# 2) 테스트용 가상 키가 입력된 상태로 버튼 조건 확인
# 실제 API 호출은 하지 않습니다.
ready = ai_summary_blockers(paper, True, evidence, "sk-example-only")
print("차단 원인:", ready)
```

```python
# 검증 셀 2: 입력 값이 준비되면 버튼 활성화 가능
assert ready == []
print("검증 2 통과: 버튼을 활성화할 수 있습니다. 실제 인증 여부는 별도 확인")
```

```python
# 3) 초록/PDF 내용이 없거나 확인하지 않으면
missing = ai_summary_blockers(paper, False, "", "sk-example-only")
print("차단 원인:", missing)
```

```python
# 검증 셀 3: 본문·동일 논문 확인이 필요
assert missing == ["EVIDENCE", "CONFIRM"]
print("검증 3 통과")
```
