# 자료종류별 참고문헌 표기 실습

**과제용 표기 3종**(학술지 논문, 석사·박사 학위논문, 기관 연구보고서)의 예시를 직접 생성합니다. 인증키와 인터넷이 필요하지 않습니다. Python 코드는 `citation_styles.py`가 같은 폴더에 있을 때 실행됩니다.

```python
from citation_styles import (
    JOURNAL, THESIS, REPORT, THESIS_EXAMPLE,
    complete_citation, format_reference, missing_citation_fields,
)

# 실제 사용자께서 제공한 학위논문 서지 예시
paper = {
    "자료명/저자사항": "Amazon RDS 데이터베이스 디지털포렌식 조사 기법 연구 /김수용",
    "DB": "학위논문",
    "발행년": "2026",
}

citation = complete_citation(paper, {
    "type": THESIS,
    "degree": "석사학위논문",
    "university": "고려대학교 대학원",
    "publication_date": "2026년 8월",
})

print("[학위논문 참고문헌]")
print(format_reference(citation))
```

### 검증 셀 1 — 학위논문 표기 일치

```python
assert citation["type"] == THESIS
assert format_reference(citation) == THESIS_EXAMPLE
assert missing_citation_fields(citation) == []
print("검증 통과: 학위논문 형식이 예시와 일치합니다.")
```

```python
# 아래 값은 표기형식을 확인하기 위한 가상 예시 데이터입니다.
journal = complete_citation({}, {
    "type": JOURNAL,
    "author": "예시저자",
    "title": "예시 연구",
    "journal": "예시 학술지",
    "volume": "제3권",
    "issue": "제2호",
    "pages": "12-25",
    "publication_date": "2026년 10월",
    "doi": "10.0000/example",
})

report = complete_citation({}, {
    "type": REPORT,
    "author": "예시연구기관",
    "title": "예시 보고서",
    "publisher": "예시연구기관",
    "report_number": "EX-2026-01",
    "publication_date": "2026년 10월",
    "url": "https://example.org/report",
})

print("[학술지 논문]")
print(format_reference(journal))
print("\n[기관 연구보고서]")
print(format_reference(report))
```

### 검증 셀 2 — 자료종류별 필수 요소 확인

```python
assert "예시 학술지" in format_reference(journal)
assert "DOI: 10.0000/example" in format_reference(journal)
assert "EX-2026-01" in format_reference(report)
assert "https://example.org/report" in format_reference(report)
assert missing_citation_fields(journal) == []
assert missing_citation_fields(report) == []
print("검증 통과: 학술지 논문·기관 연구보고서 형식이 정상입니다.")
```

```python
# 빠진 정보는 만들어내지 않고 표시해야 합니다.
incomplete = complete_citation({"자료명/저자사항": "제목만 있는 논문", "DB": "국내학술기사"})
print(format_reference(incomplete))
print("확인할 항목:", ", ".join(missing_citation_fields(incomplete)))
```

### 검증 셀 3 — 누락된 DOI 등을 임의로 생성하지 않음

```python
assert "[DOI 미확인]" in format_reference(incomplete)
assert "DOI" in missing_citation_fields(incomplete)
assert "[저자 미확인]" in format_reference(incomplete)
print("검증 통과: 확인되지 않은 정보는 미확인이라고 표시됩니다.")
```
