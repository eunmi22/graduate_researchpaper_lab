# v1.9 외부 검색 링크 만들기 실습
이 노트북은 **검색 링크 주소 생성**을 연습하고 별도 `assert` 셀로 검증합니다. 실제 학술정보 API 호출이나 사용자 인증키는 사용하지 않습니다.

> 주의: Google 사이트 검색은 실제로 검색 결과를 열지만 결과를 이 노트북에 자동 수집하는 API가 아닙니다.


```python
# 셀 1: 대학 dCollection 링크 생성
from dcollection_fallback import make_dcollection_destinations

대학 = "세종사이버대학교"
검색어 = "내부통제"
대학_링크 = make_dcollection_destinations(대학, 검색어)

print("대학 dCollection:", 대학_링크["repository_url"])
print("학교 자료검색 화면:", 대학_링크["repository_search_url"])
print("키워드 웹검색 결과:", 대학_링크["web_search_url"])
```

```python
# 검증 셀 1: 실제 학교 주소와 검색어 포함 여부
from urllib.parse import urlparse, parse_qs

assert 대학_링크["repository_url"] == "https://sjcu.dcollection.net"
assert 대학_링크["repository_search_url"].endswith("/srch/srchSearch")
검색_쿼리 = parse_qs(urlparse(대학_링크["web_search_url"]).query)["q"][0]
assert "site:sjcu.dcollection.net" in 검색_쿼리
assert "내부통제" in 검색_쿼리
print("대학별 링크 검증 통과!")
```

```python
# 셀 2: 학회·협회·보고서 검색 링크 생성
from external_research import research_search_destination, research_search_table
import pandas as pd

자료출처 = "한국중견기업연합회"
검색어 = "내부통제"
기관_링크 = research_search_destination(자료출처, 검색어)
print("외부 검색 결과:", 기관_링크["search_url"])

# 전체 기관 링크 비교표
결과표 = pd.DataFrame(research_search_table(검색어))
display(결과표)
```

```python
# 검증 셀 2: 키워드·기관 도메인 확인
쿼리 = parse_qs(urlparse(기관_링크["search_url"]).query)["q"][0]
assert "site:fomek.or.kr" in 쿼리
assert "내부통제" in 쿼리
assert len(결과표) >= 7
assert 결과표["키워드 검색 결과"].str.startswith("https://").all()
print("학회·협회·보고서 검색 링크 검증 통과!")
```
